```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import sys
sys.path.append("..")

from dataset_loader import TORGO, ParameterizedTORGO
```

```python
def plot_length_histogram(csv_path, bins=50):
    df = pd.read_csv(csv_path)

    # Drop rows with missing or NaN lengths
    df = df.dropna(subset=['length_seconds'])

    plt.figure(figsize=(10, 6))
    plt.hist(df['length_seconds'], bins=50, color='skyblue', edgecolor='black', range=(0, 30))
    plt.xlabel('Audio Length (seconds)')
    plt.ylabel('Number of Files')
    plt.title('Histogram of Audio Lengths (in seconds)')
    plt.grid(True)
    plt.tight_layout()
    plt.show()

# Example usage
plot_length_histogram("torgo_nopreemph_hamming256_lengths.csv")
```

```python
def count_unique_lengths(csv_path):
    df = pd.read_csv(csv_path)
    df = df.dropna(subset=['length_seconds'])

    unique_lengths = df['length_seconds'].nunique()
    print(f"Number of unique audio lengths (in seconds): {unique_lengths} max:{np.max(df['length_seconds'])} min:{np.min(df['length_seconds'])}")
    te = df['length_seconds'][df['length_seconds'] <= 5]
    print(f"te: {len(te)}")

    # Optionally, display a few of them:
    print("\nSample unique values:")
    print(df['length_seconds'].sort_values(ascending=False).unique()[:20])

# Example usage
count_unique_lengths("torgo_nopreemph_hamming256_lengths.csv")
```

```python
csv_path = 'torgo_nopreemph_hamming256_lengths_prompts_categories.csv'

# Load the full DataFrame
df = pd.read_csv(csv_path)

# Count non-null and null values in the 'prompt_category' column
non_empty_count = df['prompt_category'].notna().sum()
empty_count = df['prompt_category'].isna().sum()
# Count how many rows have 'undefined' in 'prompt_category'
undefined_count = (df['prompt_category'] == 'undefined').sum()
restricted_sentence_count = (df['prompt_category'] == 'restricted_sentence').sum()
unrestricted_sentence_count = (df['prompt_category'] == 'unrestricted_sentence').sum()
short_word_count = (df['prompt_category'] == 'short_word').sum()
non_word_count = (df['prompt_category'] == 'non_word').sum()

print(f"Rows with a value in 'prompt_category': {non_empty_count}")
print(f"Rows without a value in 'prompt_category': {empty_count}")
print(f"Rows with 'undefined' in 'prompt_category': {undefined_count}")
print(f"Rows with 'restricted_sentence' in 'prompt_category': {restricted_sentence_count}")
print(f"Rows with 'unrestricted_sentence' in 'prompt_category': {unrestricted_sentence_count}")
print(f"Rows with 'short_word' in 'prompt_category': {short_word_count}")
print(f"Rows with 'non_word' in 'prompt_category': {non_word_count}")
```
